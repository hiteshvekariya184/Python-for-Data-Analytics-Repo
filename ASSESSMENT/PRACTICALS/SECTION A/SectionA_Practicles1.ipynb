# Dilevery App using class

```python
class Delivery:
    def __init__(self, rider_name, location, order_id):   #_int_ function is used to initialize the attributes of the class
        self.rider_name = rider_name                        # And self is used to refer to the instance of the class
        self.location = location
        self.order_id = order_id
        self.status = "On the way"

    def update_location(self, new_location):
        self.location = new_location

    def complete_delivery(self):
        self.status = "Completed"


# Create a delivery
delivery1 = Delivery(
    "John Doe",
    "Main Street, Cityville",
    "ORD12345"
)

delivery2 = Delivery("Amit", "Surat", 102)
delivery3 = Delivery("Priya", "Rajkot", 103)

print(delivery1.rider_name)
print(delivery1.location)
print(delivery1.order_id)
print(delivery1.status)

# Update GPS location
delivery1.update_location("Satellite Road, Ahmedabad")

# Mark delivery as complete
delivery1.complete_delivery()

print("\nUpdated Delivery:")
print("Location:", delivery1.location)
print("Status:", delivery1.status)
```

# handling of exception Try-except program

```python
while True:
    try:
        amount = float(input("Enter the order amount: "))

        print("Order amount is:", amount)
        break

    except ValueError:
        print("Invalid input! Please enter a numeric value.")  # if user enter non-numeric value then this message will be displayed.
```
