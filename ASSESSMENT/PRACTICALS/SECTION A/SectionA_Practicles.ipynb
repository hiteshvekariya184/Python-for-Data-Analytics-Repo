# Difference between List of Tuples & Dictonary
##### Difference comes at the time of searching the data

```python
# Dictonary
menu = {
    'Paneer Burger': 180,
    'Masala Dosa': 90
}

print("(Search through Dictionary) Price of Paneer Burger is : ",menu['Paneer Burger'])
```

```python
# List of Tuples
menu = [
    ('Paneer Burger', 180, 'Snacks'),
    ('Masala Dosa', 90, 'Breakfast')
]

for dish, price, category in menu:
    if dish == 'Paneer Burger':
        print("(Search through List of Tuples) Price of Paneer Burger is : ",price)
```

# Calculation of delivery fees using funcion

```python
# Delivery Orders
orders = [
    {"customer": "Hitesh", "distance": 8},
    {"customer": "Amit", "distance": 5},
    {"customer": "Priya", "distance": 2}
]

print("Orders before sorting:")
for order in orders:
    print(order)

# Lambda function to sort orders by distance
orders.sort(key=lambda order: order["distance"])

print("\nOrders sorted by distance:")
for order in orders:
    print(order)
```

```python
# Using def keyword for function declaration
def calculate_delivery_fee(order_value, distance):
    if order_value >= 500:
        return 0
    elif distance <= 5:
        return order_value + 30
    else:
        return order_value + 60

calculated_fee = calculate_delivery_fee(400, 60)
print("Calculated delivery fees of : ", calculated_fee)
```
