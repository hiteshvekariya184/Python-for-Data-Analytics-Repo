## Tasks_04 Delivery Rider OOP System

```python
import csv
import os


# -----------------------------
# Rider Class
# -----------------------------

class Rider:

    def __init__(self, rider_id, name, status="Available", total_deliveries=0):
        self.rider_id = rider_id
        self.name = name
        self.status = status
        self.total_deliveries = total_deliveries

    def assign_order(self, order_id):
        self.status = "On Delivery"
        print(
            f"Order {order_id} assigned to Rider "
            f"{self.rider_id} - {self.name}"
        )
        print("Rider status: On Delivery")

    def complete_delivery(self):
        self.total_deliveries += 1
        self.status = "Available"

        print(f"Delivery completed by {self.name}.")
        print(f"Total deliveries: {self.total_deliveries}")
        print("Rider status: Available")

    def display_info(self):
        print("-" * 35)
        print(f"Rider ID          : {self.rider_id}")
        print(f"Name              : {self.name}")
        print(f"Status            : {self.status}")
        print(f"Total Deliveries  : {self.total_deliveries}")
        print("-" * 35)


# -----------------------------
# Load Riders from CSV
# -----------------------------

def load_riders():
    riders = []

    if os.path.exists("riders.csv"):

        with open("riders.csv", "r", newline="") as file:

            reader = csv.DictReader(file)

            for row in reader:
                rider = Rider(
                    row["rider_id"],
                    row["name"],
                    row["status"],
                    int(row["total_deliveries"])
                )

                riders.append(rider)

        print("Rider data loaded from riders.csv.")

    else:
        # Create at least three riders on first run
        riders.append(Rider("R001", "Rahul"))
        riders.append(Rider("R002", "Amit"))
        riders.append(Rider("R003", "Hitesh"))

        print("No riders.csv found.")
        print("Created 3 default riders.")

    return riders


# -----------------------------
# Save Riders to CSV
# -----------------------------

def save_riders(riders):

    with open("riders.csv", "w", newline="") as file:

        fieldnames = [
            "rider_id",
            "name",
            "status",
            "total_deliveries"
        ]

        writer = csv.DictWriter(
            file,
            fieldnames=fieldnames
        )

        writer.writeheader()

        for rider in riders:

            writer.writerow({
                "rider_id": rider.rider_id,
                "name": rider.name,
                "status": rider.status,
                "total_deliveries": rider.total_deliveries
            })

    print("\nRider data saved successfully to riders.csv.")


# -----------------------------
# Find Rider by ID
# -----------------------------

def find_rider(riders, rider_id):

    for rider in riders:

        if rider.rider_id == rider_id:
            return rider

    return None


# -----------------------------
# Main Program
# -----------------------------

def main():

    riders = load_riders()

    while True:

        print("\n========== DELIVERY RIDER SYSTEM ==========")
        print("1. Display All Riders")
        print("2. Assign Order")
        print("3. Complete Delivery")
        print("4. Exit")

        choice = input("Enter your choice: ")

        # -------------------------
        # Display Riders
        # -------------------------

        if choice == "1":

            print("\n========== RIDER DETAILS ==========")

            for rider in riders:
                rider.display_info()

        # -------------------------
        # Assign Order
        # -------------------------

        elif choice == "2":

            rider_id = input("Enter Rider ID: ")
            order_id = input("Enter Order ID: ")

            rider = find_rider(riders, rider_id)

            if rider:

                if rider.status == "Available":
                    rider.assign_order(order_id)
                else:
                    print(
                        f"Rider {rider_id} is currently "
                        f"{rider.status}."
                    )

            else:
                print("Rider not found.")

        # -------------------------
        # Complete Delivery
        # -------------------------

        elif choice == "3":

            rider_id = input("Enter Rider ID: ")

            rider = find_rider(riders, rider_id)

            if rider:

                if rider.status == "On Delivery":
                    rider.complete_delivery()
                else:
                    print(
                        "This rider does not have an active delivery."
                    )

            else:
                print("Rider not found.")

        # -------------------------
        # Exit
        # -------------------------

        elif choice == "4":

            save_riders(riders)

            print("Thank you for using the Delivery Rider System.")
            break

        else:

            print("Invalid choice. Please try again.")


# Run program
main()
```
