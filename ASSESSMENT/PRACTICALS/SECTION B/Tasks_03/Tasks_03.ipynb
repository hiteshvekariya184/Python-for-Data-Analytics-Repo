## Tasks_03 Order History File Logger

```python
import json

FILE_NAME = "orders.json"


def load_orders():
    try:
        with open(FILE_NAME, "r") as file:
            return json.load(file)

    except FileNotFoundError:
        return []


def save_orders(orders):
    with open(FILE_NAME, "w") as file:
        json.dump(orders, file, indent=4)


def add_order():
    try:
        customer_name = input("Enter customer name: ")

        items_input = input("Enter items (comma-separated): ")
        items = [item.strip() for item in items_input.split(",")]

        total_amount = float(input("Enter total amount: "))

        order_status = input("Enter order status: ")

        # Load existing orders
        orders = load_orders()

        # Create new order
        new_order = {
            "customer_name": customer_name,
            "items": items,
            "total_amount": total_amount,
            "order_status": order_status
        }

        # Add new order
        orders.append(new_order)

        # Save updated orders
        save_orders(orders)

        print("\nOrder added successfully!")


    except ValueError:
        print("\nError: Total amount must be a numeric value.")


def view_orders():
    try:
        with open(FILE_NAME, "r") as file:
            orders = json.load(file)

        if not orders:
            print("\nNo orders found.")
            return

        print("\n========== ORDER HISTORY ==========")

        for i, order in enumerate(orders, start=1):
            print(f"\nOrder {i}")
            print("-" * 30)
            print(f"Customer Name : {order['customer_name']}")
            print(f"Items         : {', '.join(order['items'])}")
            print(f"Total Amount  : ₹{order['total_amount']:.2f}")
            print(f"Order Status  : {order['order_status']}")

    except FileNotFoundError:
        print("\nNo order history found. orders.json does not exist yet.")


def main():
    while True:
        print("\n========== FOOD DELIVERY SYSTEM ==========")
        print("1. Add Order")
        print("2. View Orders")
        print("3. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_order()

        elif choice == "2":
            view_orders()

        elif choice == "3":
            print("Thank you!")
            break

        else:
            print("Invalid choice. Please try again.")


main()
```
