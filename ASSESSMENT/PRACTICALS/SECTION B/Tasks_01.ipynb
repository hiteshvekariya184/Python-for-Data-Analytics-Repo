# Task 1: Delivery Fee Calculator 

```python
# Get input from the user
Order_Value = float(input("Enter the order value: "))
Distance_of_Delivery = float(input("Enter the distance of delivery: "))

# Check for negative values
if Order_Value < 0 or Distance_of_Delivery < 0:
    print("Error: Order value and delivery distance cannot be negative.")
else:
    # Calculate delivery fee
    if Order_Value >= 500:
        delivery_fee = 0
    elif Distance_of_Delivery <= 5:
        delivery_fee = 30
    else:
        delivery_fee = 60

    # Calculate final amount
    final_amount = Order_Value + delivery_fee

    # Display bill
    print("\n----- Delivery Bill -----")
    print(f"Item Total       : Rs {Order_Value:.2f}")
    print(f"Delivery Fee     : Rs {delivery_fee:.2f}")
    print(f"Final Amount     : Rs {final_amount:.2f}")
```
