# Task 2: Restaurant Menu Manager

```python
# Task 2: Restaurant Menu Manager

# Restaurant menu
menu = {
    "Pizza": {
        "price": 250,
        "category": "Main Course"
    },
    "Burger": {
        "price": 150,
        "category": "Main Course"
    },
    "Pasta": {
        "price": 200,
        "category": "Main Course"
    },
    "French Fries": {
        "price": 100,
        "category": "Starter"
    },
    "Cold Coffee": {
        "price": 120,
        "category": "Beverage"
    },
    "Ice Cream": {
        "price": 80,
        "category": "Dessert"
    }
}


# Function 1: View all menu items
def view_menu():
    print("\n----------- MENU OF RESTAURANT -----------")
    print(f"{'No.':<5}{'Dish Name':<20}{'Price':<10}{'Category'}")
    print("-" * 50)

    number = 1

    for dish, details in menu.items():
        print(f"{number:<5}{dish:<20}Rs {details['price']:<7}{details['category']}")
        number += 1


# Function 2: Filter menu by category
def filter_category():
    category = input("\nEnter category: ")

    print(f"\nItems in category: {category}")
    found = False

    for dish, details in menu.items():
        if details["category"].lower() == category.lower():
            print(f"{dish} - Rs {details['price']}")
            found = True

    if not found:
        print("No items found in this category.")


# Function 3: Search for a dish
def search_dish():
    dish_name = input("\nEnter dish name: ")

    found = False

    for dish, details in menu.items():
        if dish.lower() == dish_name.lower():
            print(f"\nDish: {dish}")
            print(f"Price: Rs {details['price']}")
            found = True
            break

    if not found:
        print("Dish not found.")


# Main loop
while True:

    print("\n===== Restaurant Menu Manager =====")
    print("1. View All Items")
    print("2. Filter Items by Category")
    print("3. Search for a Dish")
    print("0. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        view_menu()

    elif choice == "2":
        filter_category()

    elif choice == "3":
        search_dish()

    elif choice == "0":
        print("Thank you! Program ended.")
        break

    else:
        print("Invalid choice. Please try again.")
```
