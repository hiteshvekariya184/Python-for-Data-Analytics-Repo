## Create a Song Class

```python
class Song: 
    def __init__(self, title, artist, duration): 
        self.title = title 
        self.artist = artist 
        self.duration = duration

song1 = Song("Shape Of You", "Ed Sheeran", 233)
print(f"Title: {song1.title}, Artist: {song1.artist}, Duration: {song1.duration} seconds")
```
