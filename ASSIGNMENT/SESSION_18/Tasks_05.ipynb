## Flipkart-Style Order ID

```python
import re 

pattern = r"\bOD\d{18}\b" 
orders = [ "Order ID: OD123456789012345000", "Order ID: OD987654321098765432" ] 

for order in orders: 
    result = re.search(pattern, order) 
    if result: 
        print("Valid Order ID:", result.group())
    else: 
        print("Invalid Order ID")
```
