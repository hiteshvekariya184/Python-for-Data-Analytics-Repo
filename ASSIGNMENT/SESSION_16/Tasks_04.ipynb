## Instagram Post Generator

```python
def insta_posts_generator(posts): 
    for post in posts: 
        yield post 

posts = [ 
    "Weekend vibes!", 
    "New memories!", 
    "Travel diary!", 
    "Good food, good mood!" 
] 

generator = insta_posts_generator(posts) 

try: 
    while True: print(next(generator)) 
except StopIteration: 
    print("All posts printed.")
```
