## Printing of Datas and it's type

```python
product_name ="Smartphone"
price = 699.99
is_available = True

print(f"The product name is {product_name}.")
print("The type of the product is : ",type(product_name))

print(f"\nThe price of the product is ${price}.")
print("The type of the product price is : ",type(price))

print(f"\nIs the product available? {is_available}.")
print("The type of the product availability is : ",type(is_available))
```
