```python
# Spotify Playlist Recommendation System

"""
This program demonstrates how a simple Spotify
playlist recommendation system might work.

The system can:
1. Check the user's favorite music genre.
2. Check the user's preferred artist.
3. Recommend songs based on these preferences.
"""

# Store the user's favorite genre
genre = "Pop"

# Store the user's favorite artist
artist = "Taylor Swift"

# Display the user's preferences
print("Favorite Genre:", genre)
print("Favorite Artist:", artist)

# Recommend a playlist based on the genre
if genre == "Pop":
    print("Recommended Playlist: Today's Pop Hits")

elif genre == "Rock":
    print("Recommended Playlist: Rock Classics")

else:
    print("Recommended Playlist: Discover Weekly")
```
