## FoodOrder and ZomatoOrder

```python
class FoodOrder: 
    def __init__(self, price): 
        self.price = price 
    
    def calculate_total(self): 
        return self.price

class ZomatoOrder(FoodOrder): 
    def calculate_total(self): 
        return self.price * 1.05

order = ZomatoOrder(1000) 
print("Total:", order.calculate_total())
```
