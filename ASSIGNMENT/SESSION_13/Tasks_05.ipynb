## Lifetime of Local and Global Variables

```python
app_status = "Offline" 

def check_status(): 
    user_status = "Online" 
    print("During Function - User:", user_status) 
    print("During Function - App:", app_status) 
    print("Before Function - App:", app_status) 

check_status() 
print("After Function - App:", app_status)
```
