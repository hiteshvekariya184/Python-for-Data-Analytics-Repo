## Recursive Instagram Likes

```python
def count_likes(posts): 
    total = 0 
    for post in posts: 
        total += post["likes"] 
        if "replies" in post: 
            total += count_likes(post["replies"]) 
    return total 

posts = [ 
    { 
        "likes": 100, 
        "replies": [ 
            {"likes": 20}, 
            {"likes": 30} 
        ] 
    },
    { 
        "likes": 50, 
        "replies": [ 
            { 
                "likes": 10, 
                "replies": [ 
                    {"likes": 5} 
                ] 
            } 
        ] 
    } 
] 

print("Total Likes:", count_likes(posts))
```
