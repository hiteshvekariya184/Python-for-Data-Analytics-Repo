## Playlist with Encapsulation

```python
class Playlist: 
    def __init__(self): 
        self._songs = [] 
    
    def add_song(self, song): 
        self._songs.append(song) 
    
    def remove_song(self, song): 
        if song in self._songs: 
            self._songs.remove(song) 
    
    def get_songs(self): 
        return self._songs 

playlist = Playlist() 
playlist.add_song("Perfect") 
playlist.add_song("Shape of You") 
playlist.remove_song("Perfect") 
print("Songs:", playlist.get_songs())
```
