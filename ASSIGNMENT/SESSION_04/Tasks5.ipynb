## Cleaning of List of extra spaces

```python
product_names = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']

cleaned_product_names = []

for name in product_names:
    names = name.strip()  # Remove extra spaces
    names = names.replace("-", " ")  # Replace hyphens with spaces
    names = names.title()  # Capitalize the first letter of each word

    cleaned_product_names.append(names)

print(cleaned_product_names)
```
