## Using of Split() & Slice()

```python
string = 'Apple iPhone 14 Pro Max'

string_split = string.split(' ')
print(string_split)

print("Brand Name : ",string_split[0]) # Output: "Apple"

string_split = string_split[1:]
print("Model Name :" , " ".join(string_split)) # Output: "iPhone 14 Pro Max"
```
