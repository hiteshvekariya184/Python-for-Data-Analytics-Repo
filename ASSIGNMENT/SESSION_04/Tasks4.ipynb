### Bulidng of Function which shows the result as formated string

```python
def format_product_display(name, price):
    formatted_name = name.title()  # Capitalize the first letter of each word
    formatted_price = f"Rs {price:.2f}"  # Format price to two decimal places with a dollar sign

    return f"Formated Date is : {formatted_name} - {formatted_price}"

Result = format_product_display("Boat Earbuds", 1299)
print(Result)  # Output: "Boat Earbuds - Rs 1299.00"
```
