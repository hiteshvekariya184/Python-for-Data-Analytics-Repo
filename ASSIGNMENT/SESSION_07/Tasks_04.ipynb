## Nested if-else Program

```python
# Take input from the user
cart_value = float(input("Enter your Flipkart cart value: ₹"))
payment_method = input("Enter payment method (UPI, Card, Cash): ")

# Nested if statements
if cart_value > 1000:
    if payment_method == "UPI":
        print(f"Your cart value is ₹{cart_value} & payment method is {payment_method}.So, You are eligible for 10% cashback.")
    else:
        print(f"Your cart value is ₹{cart_value} & payment method is {payment_method}.So, You are eligible for 5% cashback.")
else:
    print(f"Your cart value is ₹{cart_value} & payment method is {payment_method}.So, There is No cashback.")
```
