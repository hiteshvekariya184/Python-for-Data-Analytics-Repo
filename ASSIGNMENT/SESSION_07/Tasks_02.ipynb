## if,elif,else program

```python
followers = int(input("Enter the number of followers: "))

if followers < 10000:
    print(f"You are Micro Follower with {followers} followers.")
elif followers < 100000:
    print(f"You are a Rising Star Influencer with {followers} followers.")
else:
    print(f"You are a Celebrity Influencer with {followers} followers.")
```
