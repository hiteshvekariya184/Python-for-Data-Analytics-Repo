## if-else Statement

```python
# Ask the user for their Zomato order total
total = float(input("Enter your Zomato order total: ₹"))

# Check the order total
if total > 299:
    print(f"Your order total is ₹{total}. You are eligible for free delivery.")
elif total >= 200:
    print(f"Your order total is ₹{total}. Add more items for free delivery.")
else:
    print(f"Your order total is ₹{total}. Delivery charges apply.")
```
