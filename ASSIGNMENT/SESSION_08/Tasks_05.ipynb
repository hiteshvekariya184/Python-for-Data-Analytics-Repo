## Looping Concept with if-elif-else concept Program 

```python
# Instagram follower counts
followers = [120, 1500, 23000, 800, 45000]

# Classify each account based on follower count
for count in followers:
    if count < 1000:
        print(count, "- Micro")
    elif count <= 10000:
        print(count, "- Influencer")
    else:
        print(count, "- Celebrity")
```
