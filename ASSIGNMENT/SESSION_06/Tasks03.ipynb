## Removing of list from dictionary

```python
# Dictionary of Spotify playlist prices
playlist_prices = {
    "Daily Mix": 99,
    "Chill Vibes": 149,
    "Workout Hits": 249,
    "Bollywood Beats": 129,
    "Rock Classics": 179
}

# Remove a playlist using del
del playlist_prices["Chill Vibes"]

# Print the dictionary after deletion
print(playlist_prices)
```
