## Function of Updating the dictionary

```python
# Dictionary of Spotify playlist prices
playlist_prices = {
    "Daily Mix": 99,
    "Chill Vibes": 149,
    "Workout Hits": 199,
    "Bollywood Beats": 129,
    "Rock Classics": 179
}

# Function to update playlist price
def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

# Update the price of one playlist
update_playlist_price("Workout Hits", 249)

# Print the updated dictionary
print(playlist_prices)
```
