## append and pop method using on list

```python
customer_ids = [101, 102, 103]

# Add one customer
customer_ids.append(104)

# Add multiple customers
customer_ids.extend([105, 106])

# Remove last customer
customer_ids.pop()

# Print updated list
print("Updated Customer IDs:", customer_ids)

# Iterate through list
for customer_id in customer_ids:
    print("Customer ID:", customer_id)
```
