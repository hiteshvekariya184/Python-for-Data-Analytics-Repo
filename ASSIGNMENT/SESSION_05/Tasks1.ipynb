### Creation of Playlist and then adding new songs through append() method

```python
Playlists_id = [101 , 102 , 103 , 104 , 105]

print("The old playlist ID is:", Playlists_id)

#Update the playlist ID
print("\nUpdating the playlist ID...")
Playlists_id.append(106)
Playlists_id.append(108)
print("The new playlist ID is:", Playlists_id)
```
