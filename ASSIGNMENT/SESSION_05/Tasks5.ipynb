### Difference B/W List and Tuple

```python
# User's favorite genres may change, so use a list
favorite_genres = ["Rock", "Pop", "Hip-Hop"]

# IRCTC train classes are a fixed set, so use a tuple
train_classes = ("Sleeper", "AC 3 Tier", "AC 2 Tier")

print("Favorite Genres:", favorite_genres)
print("Train Classes:", train_classes)


# A list is mutable, which means its values can be changed after creation.
#     in list we can use append(), extend(), insert(), remove(), pop(), clear(), sort(), reverse() methods to modify the list.
# A tuple is immutable, which means its values cannot be changed after creation.
#     but in tuple we can use count() and index() methods to access the values of the tuple.
```
