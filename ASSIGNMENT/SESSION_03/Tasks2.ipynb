## Programing of billing of order using type casting

```python
# Take order price as a string
price = input("Enter the Zomato order price (Rs): ")

# Convert string to float
price = float(price)

# Calculate 18% GST
gst = price * 18 / 100

# Calculate final bill
final_amount = price + gst

# Display the bill
print("\n----- Zomato Bill -----")
print(f"Order Price : Rs {price:.2f}")
print(f"GST (18%)   : Rs {gst:.2f}")
print(f"Final Bill  : Rs {final_amount:.2f}")
```
