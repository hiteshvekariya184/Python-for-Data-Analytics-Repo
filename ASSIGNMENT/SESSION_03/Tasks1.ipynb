## Printing of Datas and it's Datatype

```python
number_of_followers = 1000
average_rating = 4.5
favorite_app = "Youtube"
is_premium_user = True

print("Number of followers:", number_of_followers)
print("Type of number_of_followers:", type(number_of_followers))

print("\nAverage rating:", average_rating)
print("Type of average_rating:", type(average_rating))

print("\nFavorite app:", favorite_app)
print("Type of favorite_app:", type(favorite_app))

print("\nIs premium user:", is_premium_user)
print("Type of is_premium_user:", type(is_premium_user))
```
