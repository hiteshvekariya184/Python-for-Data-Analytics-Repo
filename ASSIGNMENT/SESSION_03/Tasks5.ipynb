#### Finding of max rating from list of string of rating using loop or list comprehension

```python
List_of_Ratings =  ['4.5', '3.0', '5', '4.2']

for rating in List_of_Ratings:      #For loop to convert the ratings from string to float which is known as type casting.
    try:
        # Convert the rating to a float
        numeric_rating = float(rating)
        print(f"Rating: {numeric_rating}")
    except ValueError:
        print(f"Invalid rating value: {rating}")

Maximum_Rating = max(float(rating) for rating in List_of_Ratings)

#List Comprehension to find the maximum rating from the list of ratings
print(f"\nMaximum rating from the List of Ratings: {List_of_Ratings} is : {Maximum_Rating}") 
```
