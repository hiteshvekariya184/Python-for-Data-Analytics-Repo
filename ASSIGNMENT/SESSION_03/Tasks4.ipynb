### Function which shows is disount is available or not on the basis of amount


```python
def is_discount_applicable(order_amount):
    discount_threshold = 500.0  # Example threshold for discount eligibility
    return order_amount >= discount_threshold

print(f"Is discount applicable for order amount 450.0? {is_discount_applicable(450.0)}")  # Example usage of the function
print(f"Is discount applicable for order amount 700.0? {is_discount_applicable(700.0)}")  # Example usage of the function
```
