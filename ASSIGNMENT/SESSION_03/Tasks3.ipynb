## Convertation of List of strings into float values

```python
Product_value = ['199.99', '299.50', '150.00']

for value in Product_value:
    float_value = float(value)
    print(f"The float value of {value} is {float_value}")
    print(f"Type of element {value} is {type(float_value)}\n")

Total_value = sum(float(value) for value in Product_value)
print(f"The total value of the products is: {Total_value}")
```
