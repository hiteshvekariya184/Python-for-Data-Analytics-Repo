## Convertation of numeric string into float

```python
sales = "12500.50"
print(type(sales))

#Convert the string to a float this is known as type casting
sales_float = float(sales)
print(type(sales_float))

#Operation on the float value
total_sales = sales_float * 1.10  # Assuming a 10% increase in sales
print(f"Total sales after increase of 10% : {total_sales:.2f}")
```
